```python
import requests
from IPython.display import Markdown, display
import tiktoken
```

```python
def fetch_url(target_url: str) -> dict:
    try:
        reader_url = f"https://r.jina.ai/{target_url}"
        response = requests.get(
            reader_url,
            headers={"Accept": "application/json"},
            timeout=30
        )
        response.raise_for_status()

        # Extract 'data' field from nested response
        result = response.json()
        return result['data']
    except requests.exceptions.RequestException as e:
        print(f"Error: {e}")
        return None
```

```python
# Usage
data = fetch_url("https://r.jina.ai/https://arxiv.org/pdf/2306.12925")
if data:
    print(f"Title: {data['title']}")
    print(f"Tokens: {data['usage']['tokens']}")
    print(f"Content: {data['content'][:500]}...")
```

```python
import re

class SectionParser:
    def parse(self, text: str) -> dict:
        sections = {}
        current_section = "unknown"

        for line in text.split("\n"):
            if re.match(r"^#+\s", line):
                current_section = line.strip("# ").strip()
                sections[current_section] = []
            else:
                sections.setdefault(current_section, []).append(line)

        return {k: "\n".join(v) for k, v in sections.items()}



class Chunker:
    def chunk(self, section_name: str, text: str, chunk_size=1200):
        chunks = []
        words = text.split()
        for i in range(0, len(words), chunk_size):
            chunk_text = " ".join(words[i:i+chunk_size])
            chunks.append({
                "section": section_name,
                "text": chunk_text
            })
        return chunks
```

## Better: Paragraph-Aware Token Chunking

```python
import tiktoken

# Choose model encoding
encoding = tiktoken.encoding_for_model("gpt-5-mini")

def count_tokens(text: str) -> int:
    return len(encoding.encode(text))


def chunk_text_by_tokens(text, max_tokens=1500):
    tokens = encoding.encode(text)
    
    chunks = []
    for i in range(0, len(tokens), max_tokens):
        chunk_tokens = tokens[i:i+max_tokens]
        chunk_text = encoding.decode(chunk_tokens)
        chunks.append(chunk_text)
    
    return chunks


def semantic_token_chunk(
    text: str,
    max_tokens: int = 1500,
    overlap_tokens: int = 200,
):
    paragraphs = text.split("\n\n")
    
    chunks = []
    current_chunk = []
    current_tokens = 0

    for para in paragraphs:
        para_tokens = len(encoding.encode(para))

        # If paragraph itself exceeds max_tokens, force split
        if para_tokens > max_tokens:
            # Hard split large paragraph
            para_chunks = chunk_text_by_tokens(para, max_tokens)
            chunks.extend(para_chunks)
            continue

        # If adding paragraph exceeds max_tokens → flush
        if current_tokens + para_tokens > max_tokens:
            chunk_text = "\n\n".join(current_chunk)
            chunks.append(chunk_text)

            # Create overlap
            if overlap_tokens > 0:
                overlap_text = encoding.decode(
                    encoding.encode(chunk_text)[-overlap_tokens:]
                )
                current_chunk = [overlap_text]
                current_tokens = len(encoding.encode(overlap_text))
            else:
                current_chunk = []
                current_tokens = 0

        current_chunk.append(para)
        current_tokens += para_tokens

    if current_chunk:
        chunks.append("\n\n".join(current_chunk))

    return chunks


def chunk_sections(sections: dict, max_tokens=1500):
    all_chunks = []

    for section_name, section_text in sections.items():
        section_chunks = semantic_token_chunk(
            section_text,
            max_tokens=max_tokens
        )

        for chunk in section_chunks:
            all_chunks.append({
                "section": section_name,
                "text": chunk,
                "token_count": count_tokens(chunk)
            })

    return all_chunks

```
